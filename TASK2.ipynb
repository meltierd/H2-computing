```python
#Task 2.1
import random
def task2_1(filename, quantity, maximum):
    
    nums = []
    for i in range(quantity):
        nums.append(random.randint(0, maximum))
        
    with open(filename, 'w') as file:
        for num in nums:
            file.write(str(num) + '\n')

#main program
task2_1('randomnumbers.txt', 1000, 5000)
```

```python
#Task 2.2
def task2_2(list_of_integers):
    if len(list_of_integers) <= 1:
        return list_of_integers
    
    mid = len(list_of_integers) // 2
    
    left_half = task2_2(list_of_integers[:mid])
    right_half = task2_2(list_of_integers[mid:])
    
    return merge(left_half, right_half)

def merge(left, right):
    i = j = 0
    result = []
    
    while i < len(left) and j < len(right):
        if left[i] < right[j]:
            result.append(left[i])
            i += 1
        else:
            result.append(right[j])
            j += 1
            
    result.extend(left[i:])
    result.extend(right[j:])
    
    return result

#main program
print(task2_2([56, 25, 4, 98, 0, 18, 4, 5, 7, 0]))
```

```python
#Task 2.3
def task2_3(filename_in, filename_out):
    with open(filename_in, 'r') as file:
        arr = []
        for line in file:
            arr.append(int(line.strip()))
    #automatically closes file
    sortedarr = task2_2(arr)
    print(sortedarr)
    with open(filename_out, 'w') as file:
        for num in sortedarr:
            file.write(str(num) + '\n')

#main program
task2_3('randomnumbers.txt', 'sortednumbers.txt')
```
