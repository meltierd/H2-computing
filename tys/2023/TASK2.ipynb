```python
#Tasl 2.1
class QueueItem:
    def __init__(self, ID, quantity):
        self.ID = ID
        self.quantity = quantity

    def getID(self):
        return self.ID
    def getQuantity(self):
        return self.quantity

```

```python
#Task 2.2
class Queue:
    def __init__(self):
        self.queueList = [None] * 4
        self.headPointer = 0
        self.tailPointer = 0
        self.numberItems = 0

#Task 2.3
    def enqueue(self, QueueItem):
        if self.numberItems == 4:
            return False
        self.queueList[self.tailPointer] = QueueItem
        self.tailPointer = (self.tailPointer + 1) % 4
        self.numberItems += 1
        return True

#Task 2.4
    def dequeue(self):
        if self.numberItems == 0:
            return False
        deleted = self.queueList[self.headPointer]
        self.queueList[self.headPointer] = None
        self.headPointer = (self.headPointer + 1) % 4
        self.numberItems -= 1
        return [self.queueList[self.headPointer], deleted]

#Task 2.5
    def printQueue(self):
        print('id------qty---')
        print('------------')
        currindex = self.headPointer
        for i in range(self.numberItems):
            item = self.queueList[currindex]
            print(item.getID(), item.getQuantity())
            currindex = (currindex + 1) % 4
            
```

```python
#Task 2.3
q = Queue()

for i in range(5):
    id = input('Enter ID: ')
    qty = int(input('Enter Quantity: '))
    qitem = QueueItem(id,qty)
    res = q.enqueue(qitem)
    if res is True:
        print('Addition successful.')
    else:
        print('Addition unsuccessful.')
```

```python
#Task 2.4
for i in range(5):
    res = q.dequeue()
    if res is False:
        print('Removal unsuccessful.')
    else:
        print('id of new first item: ', res[0].getID())
        print('qty of new first item: ', res[0].getQuantity())
        print('id of deleted item: ', res[1].getID())
        print('qty of deleted item: ', res[1].getQuantity())
```
