```python
#Task 3.1
def task3_1(quantity_of_data):
    validUnits = {'KB': 10**3,
                  'MB': 10**6,
                  'GB': 10**9,
                  'TB': 10**12}
    
    if quantity_of_data[:-2].isdigit() is False or quantity_of_data[-2:] not in validUnits:
        return 'invalid data'
    
    basic = int(quantity_of_data[:-2]) * validUnits[quantity_of_data[-2:]]
    
    return basic

#main program
print(task3_1('8KB'))
print(task3_1('39MB'))
print(task3_1('123GB'))
print(task3_1('3TB'))
print(task3_1('abcKB'))
print(task3_1('8abc'))
print(task3_1('wfjsdnkjcxkzc'))
```

```python
#Task 3.2
def task3_2(quantity_of_data):
    validUnits3 = {'KiB': 2**10,
                  'MiB': 2**20,
                  'GiB': 2**30,
                  'TiB': 2**40}
    validUnits2 = {'KB': 10**3,
                  'MB': 10**6,
                  'GB': 10**9,
                  'TB': 10**12}
    
    if quantity_of_data[-3:] not in validUnits3:
        if quantity_of_data[:-2].isdigit() is False or quantity_of_data[-2:] not in validUnits2:
            return 'invalid data'
    
        basic = int(quantity_of_data[:-2]) * validUnits2[quantity_of_data[-2:]]
        return basic
    
    basic = int(quantity_of_data[:-3]) * validUnits3[quantity_of_data[-3:]]
    return basic

#main program
print(task3_2('2MiB'))
print(task3_2('13KiB'))
print(task3_2('342GiB'))
print(task3_2('5TiB'))
print(task3_2('8KB'))
print(task3_2('39MB'))
print(task3_2('123GB'))
print(task3_2('3TB'))
print(task3_2('abcKB'))
print(task3_2('8abc'))
print(task3_2('wfjsdnkjcxkzc'))
```

```python
#Task 3.3
def task3_3(quantity_of_data, target_unit):
    validUnits = {'KiB': 2**10,
                  'MiB': 2**20,
                  'GiB': 2**30,
                  'TiB': 2**40,
                  'KB': 10**3,
                  'MB': 10**6,
                  'GB': 10**9,
                  'TB': 10**12}
    
    if target_unit not in validUnits:
        return 'invalid data'
    
    basic = task3_2(quantity_of_data)
    if basic == 'invalid data':
        return basic

    target = basic / validUnits[target_unit]
    return target

#main program
print(task3_3('512MiB', 'GiB'))
print(task3_3('512MiB', 'iB'))
print(task3_3('hsdfnkasd', 'GiB'))
```
