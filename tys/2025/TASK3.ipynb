**Task 3.1**

Write a Python program that uses SQL to create a database with the **three** tables: `User`, `Post` and `Comment`

Define the primary and foreign keys for each table. **[6]**


````python
#Task 3.1
import sqlite3

conn = sqlite3.connect('SocialMedia.db')

conn.execute('DROP TABLE IF EXISTS User')
conn.execute('DROP TABLE IF EXISTS Post')
conn.execute('DROP TABLE IF EXISTS Comment')

conn.execute('''CREATE TABLE User(
  Username TEXT PRIMARY KEY,
  Blocked INTEGER)
  ''')
conn.execute('''CREATE TABLE Post(
  PostID TEXT PRIMARY KEY,
  PostText TEXT,
  DateTimePosted TEXT,
  Username TEXT,
  FOREIGN KEY (Username) REFERENCES User(Username))
  ''')
conn.execute('''CREATE TABLE Comment(
  CommentID TEXT PRIMARY KEY,
  PostID TEXT,
  Username TEXT,
  TheText TEXT,
  DateTimeCommented TEXT,
  FOREIGN KEY (PostID) REFERENCES Post(PostID),
  FOREIGN KEY (Username) REFERENCES User(Username))
  ''')

conn.commit()
conn.close()
````

**Task 3.2**

Write a Python program that uses SQL to insert the data from each of the **three** text files into the appropriate fields in each database table. **[4]**

Run your program. **[1]**


````python
#Task 3.2
conn = sqlite3.connect('SocialMedia.db')
with open('user.txt') as file: #file automatically closes
  for line in file:
    data = line.strip().split(',')
    # print(data)
    conn.execute('INSERT INTO User VALUES (?,?)', (data[0], data[1]))

with open('post.txt') as file: #file automatically closes
  for line in file:
    data = line.strip().split(',')
    # print(data)
    conn.execute('INSERT INTO Post VALUES (?,?,?,?)', (data[0], data[1], data[2], data[3]))

with open('comment.txt') as file: #file automatically closes
  for line in file:
    data = line.strip().split(',')
    # print(data)
    conn.execute('INSERT INTO Comment VALUES (?,?,?,?,?)', (data[0], data[1], data[2], data[3], data[4]))

conn.commit()
conn.close()
````

**Task 3.3**

Write a Python function that:

- uses SQL to search for all comments by a user who is blocked
- outputs each returned username and comment on a new line.

**[5]**

Run your program. **[1]**

Save your Jupyter Notebook for Task 3.


````python
#Task 3.3
def findBlocked():
  conn = sqlite3.connect('SocialMedia.db')

  query = '''SELECT User.Username, Comment.TheText
    FROM User
    JOIN Comment ON User.Username = Comment.Username
    WHERE User.Blocked == 1'''
  data = conn.execute(query)
  data = data.fetchall()
  # print(data)

  print('Username: Comment')
  for comment in data:
    print(comment[0], ': ', comment[1])

  conn.close()

#main program
findBlocked()
````
