```python
#Task 2.1
class Node:
    def __init__(self):
        self.data_value = -1
        self.left_node = -1
        self.right_node = -1
        
    def set_data(self, value):
        self.data_value = value
        
    def set_left(self, value):
        self.left_node = value
        
    def set_right(self, value):
        self.right_node = value
        
    def get_data(self):
        return self.data_value
    
    def get_left(self):
        return self.left_node
    
    def get_right(self):
        return self.right_node
    
```

```python
#Task 2.2
def calculate_nodes(levels):
    total = 2 ** (levels) -1
    print(f'A {levels} level number tree has {total} nodes')
    return total

#main program
print(calculate_nodes(5))
```

```python
#Task 2.3
print(calculate_nodes(1))
print(calculate_nodes(3))
print(calculate_nodes(5))
print(calculate_nodes(10))
```

```python
#Task 2.4
def create_empty_tree(list_of_nodes, levels):
    numNodes = int(calculate_nodes(levels))
    for i in range(numNodes):
        newNode = Node()
        list_of_nodes.append(newNode)
    return list_of_nodes

#main program
nodes = create_empty_tree([], 3)
```

```python
#Task 2.5
def assign_leaf_nodes(list_of_nodes, levels):
    numLeafNodes = 2 ** (levels - 1)
    for i in range(numLeafNodes):
        list_of_nodes[i].set_data(i + 1)
    return list_of_nodes
        
#main program
nodes = assign_leaf_nodes(nodes, 3)
```
