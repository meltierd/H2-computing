```python
#Task 3.1
def inputPuzzle():
    sudoku = []
    for i in range(9):
        rowvalid = False
        while rowvalid == False:
            row = input('Enter row of 9: ')
            row = row.split(' ')
            if len(row) != 9:
                print('too many numbers')
                continue
            allnumvalid = True
            for num in row:
                if num.isdigit() is False:
                    print('input must be integer')
                    allnumvalid = False
                elif int(num) > 9 or int(num) < 1:
                    print('integers must be between 1 and 9')
                    allnumvalid = False
            
            if allnumvalid == False: continue
            rowvalid = True
        sudoku.append(row)
    return sudoku
```

```python
#Task 3.2
def horizontalcheck(puzzle):
    for row in range(9):
        numberssofar = []
        for col in range(9):
            if puzzle[row][col] in numberssofar:
                return row + 1
            numberssofar.append(puzzle[row][col])
    return -1

print(horizontalcheck(example1))
print(horizontalcheck(example2))
```

```python
#Task 3.3
def verticalcheck(puzzle):
    for col in range(9):
        numberssofar = []
        for row in range(9):
            if puzzle[row][col] in numberssofar:
                return col + 1
            numberssofar.append(puzzle[row][col])
    return -1


print(verticalcheck(example1))
print(verticalcheck(example2))
```
