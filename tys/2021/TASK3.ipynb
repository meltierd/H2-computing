```python
#Task 3.1
class Node:
    def __init__(self, val):
        self.val = val
        self.next = None
        
class LinkedList:
    def __init__(self, head):
        self.head = head
        
    def insert(self, integer_value):
        head = self.head #temp current head
        newNode = Node(integer_value) #init newnode
        self.head = newNode #assign new head to newnode
        newNode.next = head #assign head.next to old head
        print('inserted.')
        
    def delete(self, integer_value):
        curr = self.head
        prev = None
        while curr is not None and curr.val != integer_value :
            prev = curr
            curr = curr.next
        if curr is None:
            return None
        prev.next = curr.next
        print('deleted.')
        
    def search(self, integer_value):
        curr = self.head
        while curr.val != integer_value:
            curr = curr.next
        if curr.val == integer_value:
            return True
        return False
        
    def count(self):
        count = 0
        curr = self.head
        while curr:
            curr = curr.next
            count += 1
        return count
        
    def to_String(self):
        string = '['
        curr = self.head
        while curr:
            string = string + str(curr.val) + ', '
        string = string + ']'
        return string
    
#main program
node = Node(5)
lst = LinkedList(node)
```
