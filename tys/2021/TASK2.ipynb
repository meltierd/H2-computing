```python
#Task 2.1
def task2_1(filename):
    contents = []
    
    with open(filename, 'r') as file:
        for line in file:
            contents.append(int(line.strip()))
    #file closes automatically
    
    return contents

#main program
result = task2_1('TEN.txt')
print(result)
print(len(result))
```

```python
#Task 2.2
def task2_2(list_of_integers):
    #insertion sort
    for i in range(len(list_of_integers)):
        j = i
        while j > 0 and list_of_integers[j-1] > list_of_integers[j]:
            list_of_integers[j-1], list_of_integers[j] = list_of_integers[j], list_of_integers[j-1]
            j -= 1
    return list_of_integers

#main program
print(task2_2(task2_1('TEN.txt')))
```

```python
#Task 2.3
def task2_3(list_of_integers):
    return quicksort(list_of_integers, 0, len(list_of_integers)-1)
    
def quicksort(list_of_integers, low, high):
    if low < high:
        pivotindex = partition(list_of_integers, low, high)
        quicksort(list_of_integers, low, pivotindex - 1)
        quicksort(list_of_integers, pivotindex + 1, high)
    return list_of_integers
        
def partition(list_of_integers, low, high):
    pivot = list_of_integers[high]
    i = low - 1
    for j in range(low, high):
        if list_of_integers[j] <= pivot:
            i += 1
            list_of_integers[i], list_of_integers[j] = list_of_integers[j], list_of_integers[i]
        list_of_integers[i + 1], list_of_integers[high] = list_of_integers[high], list_of_integers[i + 1]
    return i + 1
    
#main program
print(task2_3(task2_1('TEN.txt')))
```
