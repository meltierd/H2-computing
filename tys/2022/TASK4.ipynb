```python
#Task 4.1
import sqlite3

conn = sqlite3.connect('LIBRARY2022.db')

conn.execute('DROP TABLE IF EXISTS Book')
conn.execute('DROP TABLE IF EXISTS Member')
conn.execute('DROP TABLE IF EXISTS Loan')

conn.execute("""CREATE TABLE Book(
            BookID INTEGER PRIMARY KEY,
            Title TEXT,
            Genre TEXT)
            """)
conn.execute('''CREATE TABLE Member(
MemberNumber INTEGER PRIMARY KEY,
FamilyName TEXT,
GivenName TEXT)
''')
conn.execute('''CREATE TABLE Loan(
LoanID INTEGER PRIMARY KEY,
MemberNumber INTEGER,
BookID INTEGER,
DateLoaned TEXT,
Returned TEXT,
FOREIGN KEY (MemberNumber) REFERENCES Member(MemberNumber),
FOREIGN KEY (BookID) REFERENCES Book(BookID))
''')

conn.commit()
conn.close()
```

```python
#Task 4.2
conn = sqlite3.connect('LIBRARY2022.db')
with open('BOOK.txt') as file: #file closes automatically
    for line in file:
#         print(line)
        line = line.strip().split(',')
        conn.execute('INSERT INTO Book(BookID, Title, Genre) VALUES(?, ?, ?)', (int(line[0]), line[1], line[2]))

with open('MEMBER.txt') as file: #file closes automatically
    for line in file:
#         print(line)
        line = line.strip().split(',')
        conn.execute('INSERT INTO Member VALUES(?, ?, ?)', (int(line[0]), line[1], line[2]))
        
with open('LOAN.txt') as file: #file closes automatically
    for line in file:
#         print(line)
        line = line.strip().split(',')
        conn.execute('INSERT INTO Loan VALUES(?, ?, ?, ?, ?)', (int(line[0]), int(line[1]), int(line[2]), line[3], line[4]))
        
conn.commit()
conn.close()
```

```python
#Task 4.4
import sqlite3

conn = sqlite3.connect('LIBRARY2022.db')
recs = conn.execute('''SELECT Member.FamilyName, Member.GivenName, Book.Title
FROM Member, Book, Loan
WHERE Member.MemberNumber = Loan.MemberNumber AND Book.BookID = Loan.BookID AND
Loan.Returned = "False"''')

for rec in recs:
    print(rec)
conn.close()
```
