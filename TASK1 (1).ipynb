```python
#Task 1.1
def task1_1(input_value):
    input_value = str(input_value)
    if input_value.isdigit() is False or len(input_value) != 14:
        return -1
    
    #step 1
    string = ''
    for i in range(14):
        if i % 2 != 0: #odd
            string += str(int(input_value[i]) * 2)
        else: #even
            string += str(int(input_value[i]))
    
    #step 2
    total = 0
    for num in string:
        total += int(num)
    
    #step 4
    check = 10 - (total % 10)
    
    return check
```

```python
#Task 1.2
print(task1_1('12345678901234') == 7)
print(task1_1(14576567654934) == 2)
print(task1_1('abcdjrofmxuepq') == -1)
print(task1_1('123545') == -1)
```

```python
#Task 1.3
def task1_3(input_value):
    input_value = str(input_value)
    if input_value.isdigit() is False or len(input_value) != 15:
        return False
    
    check = input_value[-1]
    val = input_value[:-1]
    
    return (task1_1(val) == int(check))

#main program
print(task1_3('123456789012347'))
print(task1_3(145765676549342))
print(not task1_3('abcdjrofmxuepqr'))
print(not task1_3('123545'))
```
